**Objetivo:** implementação manual de uma rede perceptron

**História do algoritmo:**  
Em 1943, McCulloch e Pitts formularam o primeiro modelo matemático de um neurônio artificial: uma unidade lógica capaz de implementar portas como AND e OR. O problema que resolveram foi mostrar que o cérebro poderia ser descrito como um sistema lógico, mas sem qualquer capacidade de aprendizado, já que os pesos eram fixos.

Mais tarde, Donald Hebb (1949) propôs uma regra simples de atualização de conexões, mas ainda limitada. Foi então que Frank Rosenblatt, em 1958, criou o Perceptron, o primeiro modelo capaz de aprender automaticamente os pesos a partir de exemplos. O problema que ele se propunha a resolver era ir além da lógica fixa e permitir que uma máquina conseguisse ajustar sua fronteira de decisão para classificar padrões binários linearmente separáveis.

$$
\hat{y} = 
\begin{cases}
1, & \text{se } \sum_{i=1}^{n} w_i \cdot x_i + b \geq 0 \\
0, & \text{caso contrário}
\end{cases}
$$

```python
import numpy as np
```

```python
def calc_peso(weight, col, real, pred, ln):
    """Função para ajustar os pesos"""
    
    w_fin = weight+ln*(real - pred)*col
    return w_fin
```

```python
def calc_bias(bias, predito, real, lr):
    
    """Função que calcula o bias"""

    return bias+lr*(predito - real)
```

```python
def perceptron(X, y, lr, epocs):
    max_random_w = np.max(X)
    
    real = 0 
    pred = 0
    w = np.random.uniform(-2, 2, X.shape[1])
    pesos_hist = []
    epocs = range(epocs)
    b = np.random.uniform(0, 2, 1)[0]

    pesos_hist.append(w)
    pred_hist = []

    if isinstance(X, np.ndarray):
        for epoc in epocs:
            predicoes = []

            for i, row in enumerate(X):
 
                pred_col_row = 0
                for j, col in enumerate(row):
                    pred_col_row += w[j]*col

                pred_col_row += b             
                if pred_col_row > 0:
                    pred = 1
                else:
                    pred = 0


                predicoes.append(pred)
                if pred != y[i]:
                    pesos_hist.append(w)
                    w = np.array([calc_peso(w[j], col, y[i], pred, lr) for j, col in enumerate(row)])
                    b = calc_bias(b, pred, y[i], lr)

            pred_hist.append(predicoes)
                    
    return {"pesos":w, "predicoes":predicoes, "learning_rate":lr, "pesos_hist":pesos_hist}
        
```

```python
X = np.array([
    [5, 20],
    [10, 40],
    [3, 12],
    [1, 5]
])
y = np.array([0, 0, 0, 1])
```

```python
perceptron(X, y,  2, 200)
```
