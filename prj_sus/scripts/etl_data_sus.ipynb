```python
import requests
import pandas as pd
```

```python
def _get_many_offset(offset=0):
    url = "https://apidadosabertos.saude.gov.br/vigilancia-e-meio-ambiente/sistema-de-informacao-sobre-mortalidade?"
    params = {
        "limit": 1000,
        "offset": offset
    }
    try:
        response = requests.get(url, params=params)
        if response.status_code == 200:
            return response.json()
        else:
            print(f"Erro na requisição: {response.status_code}")
            return {}
    except Exception as e:
        print(f"Exceção ao fazer requisição: {e}")
        return {}
```

```python
df = pd.DataFrame()
for i in range(1000):
    result = _get_many_offset(i)
    sim_data = result.get('sim')
    if sim_data:
        dfi = pd.DataFrame(sim_data)
        df = pd.concat([df, dfi], ignore_index=True)
    else:
        print(f"Nenhum dado retornado para offset {i}.")
```
