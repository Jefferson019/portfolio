````python
import itertools
````

```count(start=0, step=1)```: Cria um iterador que retorna números espaçados uniformemente, começando por start e incrementando por step.

````python

for i in itertools.count(10, 2):
    print(i)
    if i > 20:
        break
````

```cycle(iterable)```: Cria um **iterador** que repete os elementos de um iterável **indefinidamente**.

````python
cores = ['vermelho', 'verde', 'azul']
contador = 0
for cor in itertools.cycle(cores):
    print(cor)
    contador += 1
    if contador == 6:
        break
````

```repeat(object, times=None)```: Cria um iterador que retorna o object um número específico de vezes ou indefinidamente se a **quantidade de vezes** não for especificado.

````python
for txt in itertools.repeat('Eu sou um iterador', 3):
    print(txt)
````

```product(*iterables, repeat=1)```: Calcula o **produto cartesiano** dos iteráveis de entrada.

````python
letras = ['Mariana', 'Ana']
numeros = [1, 2, 5]
resultado = list(itertools.product(letras, numeros))
print(resultado)
````

```permutations(iterable, r=None)```: Retorna permutações de comprimento r dos elementos do iterável. Se r não for especificado, o comprimento do iterável é usado.

````python
import itertools

items = ['a', 'b', 'c']
resultado = list(itertools.permutations(items, 3))
print(resultado)
````

```combinations(iterable, r)```: Retorna combinações de comprimento r dos elementos do iterável. A ordem dos elementos não importa.

````python
items = ['a', 'b', 'c']
resultado = list(itertools.combinations(items, 2))
print(resultado)
````

```combinations_with_replacement(iterable, r)```: Semelhante ao combinations, mas permite que os elementos individuais sejam repetidos.

````python
items = ['a', 'b']
resultado = list(itertools.combinations_with_replacement(items, 2))
print(resultado)
````

```chain(*iterables)```: Encadeia múltiplos iteráveis juntos em um único iterador.

````python
lista1 = [1, 2, 3]
lista2 = ['a', 'b', 'c']
resultado = list(itertools.chain(lista1, lista2))
print(resultado)
````

```groupby(iterable, key=None)```: Agrupa elementos consecutivos que possuem o mesmo valor de chave. O iterável precisa estar ordenado pela chave para que o agrupamento funcione como esperado.

````python
dados = [('a', 1), ('a', 2), ('b', 3), ('b', 4), ('c', 5), ('e', 10), ('c', 20)]
for chave, grupo in itertools.groupby(dados, key=lambda x: x[0]):
    print(f"Chave: {chave}, Grupo: {list(grupo)}")
````

```filterfalse(predicate, iterable)```: Retorna os elementos do iterável para os quais o predicado (função) retorna False.

````python
numeros = [1, 2, 3, 4, 5, 6]
pares = list(itertools.filterfalse(lambda x: x % 2 != 0, numeros))
print(pares)
````

```takewhile(predicate, iterable)```: Retorna elementos do iterável enquanto o predicado for verdadeiro.

````python
numeros = [1, 3, 5, 2, 4, 6]
impares_iniciais = list(itertools.takewhile(lambda x: x % 2 != 0, numeros))
print(impares_iniciais)
````

```dropwhile(predicate, iterable)```: Descarta os elementos do iterável enquanto o predicado for verdadeiro e, em seguida, retorna o restante.

````python
numeros = [1, 3, 5, 2, 4, 6]
primeiro_par_e_resto = list(itertools.dropwhile(lambda x: x % 2 != 0, numeros))
print(primeiro_par_e_resto)
````

```starmap(function, iterable)```: Aplica uma função a cada item de um iterável, onde cada item já é uma tupla de argumentos.

````python
import operator

pontos = [(1, 2), (3, 4), (5, 6)]
somas = list(itertools.starmap(operator.add, pontos))
print(somas)
````

```accumulate(iterable, func=operator.add)```: Retorna uma série de resultados acumulados.

````python
numeros = [1, 2, 3, 4, 5]
somas_acumuladas = list(itertools.accumulate(numeros))
print(somas_acumuladas)

produtos_acumulados = list(itertools.accumulate(numeros, operator.mul))
print(produtos_acumulados)
````
